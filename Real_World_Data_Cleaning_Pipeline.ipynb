```python
import pandas as pd

df = pd.read_csv(r"C:\Users\MOUNAVI GOLLAPALLI\Downloads\archive (1)\AB_NYC_2019.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nFirst 5 Rows:")
display(df.head())
```

```python
missing_report = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().mean() * 100).round(2)
})

missing_report = missing_report.sort_values(
    by="Missing Count", ascending=False
)

display(missing_report)
```

```python
print("Total Rows:", len(df))
print("Duplicate Rows:", df.duplicated().sum())
print("Unique Rows:", len(df.drop_duplicates()))
```

```python
display(df.describe().round(2))
```

```python
df_clean = df.copy()

df_clean["name"] = df_clean["name"].fillna("Unknown Listing")
df_clean["host_name"] = df_clean["host_name"].fillna("Unknown Host")

df_clean["reviews_per_month"] = df_clean["reviews_per_month"].fillna(0)

print("Missing values handled successfully!")
print("Original rows:", len(df))
print("Cleaned rows:", len(df_clean))

display(df_clean.isnull().sum().to_frame("Remaining Missing Values"))
```

```python
print("Duplicate rows before cleaning:", df_clean.duplicated().sum())

print("Duplicate listing IDs:", df_clean["id"].duplicated().sum())
```

```python
print("Zero or negative prices:", (df_clean["price"] <= 0).sum())

print("Negative minimum nights:", (df_clean["minimum_nights"] < 0).sum())

print("Invalid latitude:", (~df_clean["latitude"].between(-90, 90)).sum())

print("Invalid longitude:", (~df_clean["longitude"].between(-180, 180)).sum())
```

```python
invalid_prices = df_clean[df_clean["price"] <= 0]

print("Invalid price records:", len(invalid_prices))

display(invalid_prices[[
    "id", "name", "neighbourhood_group",
    "room_type", "price", "number_of_reviews"
]])
```

```python
df_clean = df_clean[df_clean["price"] > 0].copy()

print("Invalid prices after cleaning:", (df_clean["price"] <= 0).sum())
print("Rows remaining:", len(df_clean))
print("Columns remaining:", len(df_clean.columns))
```

```python
Q1 = df_clean["price"].quantile(0.25)
Q3 = df_clean["price"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

price_outliers = df_clean[
    (df_clean["price"] < lower_limit) |
    (df_clean["price"] > upper_limit)
]

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Price outliers:", len(price_outliers))

display(price_outliers[["name", "neighbourhood_group", "room_type", "price"]].head(10))
```

```python
df_clean["price_outlier"] = (
    (df_clean["price"] < lower_limit) |
    (df_clean["price"] > upper_limit)
)

print("Outliers flagged:", df_clean["price_outlier"].sum())
print("Normal price records:", (~df_clean["price_outlier"]).sum())

display(df_clean[["name", "price", "price_outlier"]].head(10))
```

```python
df_clean["last_review"] = pd.to_datetime(
    df_clean["last_review"], errors="coerce"
)

df_clean["id"] = df_clean["id"].astype("int64")
df_clean["host_id"] = df_clean["host_id"].astype("int64")

print(df_clean.dtypes)
```

```python
quality_report = pd.DataFrame({
    "Check": [
        "Total Rows",
        "Total Columns",
        "Duplicate Rows",
        "Duplicate Listing IDs",
        "Invalid Prices",
        "Missing Values",
        "Price Outliers"
    ],
    "Result": [
        len(df_clean),
        len(df_clean.columns),
        df_clean.duplicated().sum(),
        df_clean["id"].duplicated().sum(),
        (df_clean["price"] <= 0).sum(),
        df_clean.isnull().sum().sum(),
        df_clean["price_outlier"].sum()
    ]
})

display(quality_report)
```
